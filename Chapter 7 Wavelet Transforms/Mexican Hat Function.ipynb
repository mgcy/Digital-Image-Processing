```python
import numpy as np
import math

def mexhat_np(u, v, sigma=1):
    g1 = 1 / (math.pi * sigma ** 4)
    g2 = (u ** 2 + v ** 2) / (2 * sigma ** 2) - 1
    g3 = np.exp(-(u ** 2 + v ** 2) / (2 * sigma ** 2))
    return g1 * g2 * g3

N = 100
u = np.linspace(-4, 4, N)
v = np.linspace(-4, 4, N)

u, v = np.meshgrid(u, v)

g = mexhat_np(u, v, sigma=1)
```
