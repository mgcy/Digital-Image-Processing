# Define functions

```python
import numpy as np

def exp_func(u, v):
    return 0.9 ** np.sqrt(u**2 + v**2)


def get_pho_k(M, k):
    tmp_list = list(range(0, -2 * M -1, -1))
    re = np.zeros((2*M+1, 2*M+1))

    for i in range(2*M+1):
        for j in range(2*M+1):
            re[i, j] = exp_func(tmp_list[j] + i, k)
    return re


def get_pho_0_p(M):
    tmp_list = list(range(0, -M-1, -1))

    re = np.zeros((M+1, M+1))
    for i in range(M+1):
        for j in range(M+1):
            re[i, j] = exp_func(tmp_list[j] + i, 0)
    return re


def get_pho_minus_k_p(M, k):
    tmp_list = list(range(M, -M -1, -1))
    re = np.zeros((M+1, 2*M+1))

    for i in range(M+1):
        for j in range(2*M+1):
            re[i, j] = exp_func(tmp_list[j] + i, -k)
    return re
```

```python
# @title Causal (1,1)
M=1
pho_0_p = get_pho_0_p(M)
pho_0 = get_pho_k(M, k=0)
pho_minus_1_p = get_pho_minus_k_p(M, k=1)

R = np.block([[pho_0_p, pho_minus_1_p],
              [pho_minus_1_p.T, pho_0]])

print(R.shape)

iR = np.linalg.inv(R)
sigma2 = 1 / iR[0, 0]
print('sigma^2: ', sigma2)


print("n=0")
for m in [1]:
    a = -iR[m,0]* sigma2
    print('a(m,n)=(',m,', 0 ): ', a)

print("n=1")
for m in [-1,0,1]:
    k =(2 * M + 1) + m
    a = -iR[k,0]* sigma2
    print('a(m,n)=(',m,', 1 ): ', a)
```

```python
# @title Causal (2,2)
M=2
pho_0_p = get_pho_0_p(M)
pho_0 = get_pho_k(M, k=0)

pho_mins_1 = get_pho_k(M, k=1)
pho_1 = get_pho_k(M, k=1)


pho_minus_1_p = get_pho_minus_k_p(M, k=-1)
pho_minus_2_p = get_pho_minus_k_p(M, k=-2)

R = np.block([[pho_0_p, pho_minus_1_p, pho_minus_2_p],
              [pho_minus_1_p.T, pho_0, pho_mins_1],
              [pho_minus_2_p.T, pho_1, pho_0]])
# print(R.shape)

iR = np.linalg.inv(R)
sigma2 = 1 / iR[0, 0]
print('sigma^2: ', sigma2)

print("n=0")
for m in [1, 2]:
        print('a(m,n)=(',m,', 0 ): ', -iR[m,0]* sigma2)

print("n=1")
for m in [-2, -1,0,1, 2]:
    kl = (2*M + 1) + m
    print('a(m,n)=(',m,', 1 ): ', -iR[kl,0]* sigma2)

print("n=2")
for m in [-2, -1,0,1, 2]:
    kl = 2 * (2*M + 1) + m
    print('a(m,n)=(',m,', 2 ): ', -iR[kl,0]* sigma2)
```

```python
# @title Semicausal (1,1)
M=1
pho_0 = get_pho_k(M, k=0)
pho_mins_1 = get_pho_k(M, k=-1)
pho_1 = get_pho_k(M, k=1)

R = np.block([[pho_0, pho_mins_1],
              [pho_1, pho_0]])

iR = np.linalg.inv(R)
sigma2 = 1 / iR[M, M]
print('sigma^2: ', sigma2)

for m in [-1, 0, 1]:
    for n in [0, 1]:
        kl = n * (2 * M + 1) + m + M
        print('a(m,n)=(',m,',',n, '): ', -iR[kl,M]* sigma2)
```

```python
# @title Semicausal (2,2)
M=2
pho_0 = get_pho_k(M, k=0)
pho_mins_1 = get_pho_k(M, k=-1)
pho_mins_2 = get_pho_k(M, k=-2)

pho_1 = get_pho_k(M, k=1)
pho_2 = get_pho_k(M, k=2)

R = np.block([[pho_0, pho_mins_1,pho_mins_2],
              [pho_1, pho_0,pho_mins_1],
              [pho_2, pho_1,pho_0]])

iR = np.linalg.inv(R)
sigma2 = 1 / iR[M, M]
print('sigma^2: ', sigma2)

for m in [-2, -1, 0, 1, 2]:
    for n in [0, 1, 2]:
        kl = n * (2 * M + 1) + m + M
        print('a(m,n)=(',m,',',n, '): ', -iR[kl,M] * sigma2)
```

```python
# @title Noncausal (1,1)
M = 1
pho_0 = get_pho_k(M, k=0)

pho_mins_1 = get_pho_k(M, k=-1)
pho_mins_2 = get_pho_k(M, k=-2)

pho_1 = get_pho_k(M, k=1)
pho_2 = get_pho_k(M, k=2)

R = np.block([[pho_0, pho_mins_1,pho_mins_2],
              [pho_1, pho_0,pho_mins_1],
              [pho_2, pho_1,pho_0]])

iR = np.linalg.inv(R)

k = M * (2 * M + 1) + M
print('k: ', k)
sigma2 = 1 / iR[k, k]
print('sigma^2: ', sigma2)

for m in [-1, 0, 1]:
    for n in [-1, 0, 1]:
        l = (n + M) * (2* M + 1) + m + M
        print('a(m,n)=(',m,',',n, '): ', -iR[l,k] * sigma2)
```

```python
# @title Noncausal (2,2)
M = 2
pho_0 = get_pho_k(M, k=0)

pho_mins_1 = get_pho_k(M, k=-1)
pho_mins_2 = get_pho_k(M, k=-2)
pho_mins_3 = get_pho_k(M, k=-3)
pho_mins_4 = get_pho_k(M, k=-4)

pho_1 = get_pho_k(M, k=1)
pho_2 = get_pho_k(M, k=2)
pho_3 = get_pho_k(M, k=3)
pho_4 = get_pho_k(M, k=4)

R = np.block([[pho_0, pho_mins_1,pho_mins_2,pho_mins_3,pho_mins_4],
              [pho_1, pho_0,pho_mins_1,pho_mins_2,pho_mins_3],
              [pho_2, pho_1,pho_0, pho_mins_1, pho_mins_2],
              [pho_3, pho_2,pho_1, pho_0, pho_mins_1],
              [pho_4, pho_3,pho_2, pho_1, pho_0]])

iR = np.linalg.inv(R)
ka = M * (2* M + 1) + M

sigma2 = 1 / iR[ka, ka]
print(sigma2)

for m in [-2, -1, 0, 1, 2]:
    for n in [0, 1, 2]:
        kl = (n + M)* (2* M + 1) + m + M
        print('a(m,n)=(',m,',',n, '): ', -iR[kl,ka]* sigma2)
```
