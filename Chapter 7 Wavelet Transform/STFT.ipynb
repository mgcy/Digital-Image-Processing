# Plot Filter Banks


```python
from scipy.signal import butter
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import freqz
import math

def butter_bandpass(lowcut, highcut, fs, order=9):
    return butter(order, [lowcut, highcut], fs=fs, btype='band')


# Sample rate and desired cutoff frequencies (in Hz).
fs = 12000.0
# Plot the frequency response for a few different orders.
plt.figure(1,figsize=(10,4))
plt.clf()
for i in range(5):
    b, a = butter_bandpass((i)*1000 + 500,(i)*1000 + 500 + 700, fs)
    w, h = freqz(b, a)
    plt.plot(w, abs(h), color='black')

xstick = np.linspace(0, math.pi, 4)
xstick_labels = [0, '$\\frac{\pi}{4}$', '$\\frac{\pi}{2}$', '$\pi$']
plt.xticks(xstick, xstick_labels, fontsize=14)
plt.xlabel('Normalized Frequency (Hz)')
plt.ylabel('Magnitude')
plt.grid(True)
plt.legend(loc='best')
```
