# DCT Basis Function


```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.fft import dct
import matplotlib

work_path = '/content/drive/MyDrive/DigitalImageProcessing/Chapter1/'

D = dct(np.eye(8), axis=0)

plt.figure()
plt.title('One DCT Basis Function Example')
plt.imshow(np.outer(D[:,7], D[:,7]), cmap='gray')
plt.axis('off')

linewidth = 1

bf = -20 * np.ones((64+ 7*linewidth,64 + 7*linewidth))

for i in range(8):
  for j in range(8):
    bf[i * (8+linewidth): i * (8+linewidth) + 8, j * (8+linewidth): j * (8+linewidth) + 8] = np.outer(D[:,i], D[:,j])

plt.figure()
plt.imshow(bf, cmap='gray')
plt.title('DCT Basis Function')
plt.colorbar()
plt.axis('off')

masked_array = np.ma.masked_where(bf == -20, bf)
plt.figure()
cmap = matplotlib.cm.gray  # Can be any colormap that you want after the cm
cmap.set_bad(color='blue')
plt.title('DCT Basis Function for N=8')
plt.imshow(masked_array, cmap=cmap)
plt.axis('off')

# save figure
plt.savefig(work_path+'Plots/Basis_Functions/DCT.Basis.pdf', format='pdf',
                    bbox_inches='tight', facecolor=(1, 1, 1, 0))

plt.show()
```

# Hadamard Basis Function


```python
import numpy as np
from matplotlib import pyplot as plt
import matplotlib

H1 = (1/np.sqrt(2)) * np.array([[1,1],[1,-1]])
print('H1: \n', H1)
plt.figure()
plt.title('H1')
plt.imshow(H1, cmap='gray')
plt.colorbar()

H2 = np.zeros((4,4))
for i in range(2):
  for j in range(2):
    if i==1 and j==1:
        H2[i*2:i*2+2,j*2:j*2+2] = -H1
    else:
        H2[i*2:i*2+2,j*2:j*2+2] = H1

H2 = (1/np.sqrt(2)) * H2
print('\n H2: \n', H2)
plt.figure()
plt.title('H2')
plt.imshow(H2, cmap='gray')
plt.colorbar()

H3 = np.zeros((8,8))
for i in range(2):
  for j in range(2):
    if i==1 and j==1:
        H3[i*4:i*4+4,j*4:j*4+4] = -H2
    else:
        H3[i*4:i*4+4,j*4:j*4+4] = H2

H3 = (1/np.sqrt(2)) * H3
print('\n H3: \n', H3)
plt.figure()
plt.title('H3')
plt.imshow(H3, cmap='gray')
plt.colorbar()

# plot H3 basis function

linewidth = 1

bf = -5 * np.ones((64+ 7*linewidth,64 + 7*linewidth))

for i in range(8):
  for j in range(8):
    bf[i * (8+linewidth): i * (8+linewidth) + 8, j * (8+linewidth): j * (8+linewidth) + 8] = np.outer(H3[:,i], H3[:,j])

plt.figure()
plt.imshow(bf, cmap='gray')
plt.title('Hadamard Basis Function')
plt.colorbar()

masked_array = np.ma.masked_where(bf == -5, bf)
plt.figure()
cmap = matplotlib.cm.gray  # Can be any colormap that you want after the cm
cmap.set_bad(color='blue')
plt.title('Hadamard Basis Function for N=8')
plt.imshow(masked_array, cmap=cmap)
plt.axis('off')

# save figure
plt.savefig(work_path+'Plots/Basis_Functions/Hadamard.Basis.pdf', format='pdf',
                    bbox_inches='tight', facecolor=(1, 1, 1, 0))

plt.show()
```

# Ordered Hadamard Basis Function


```python
import numpy as np
from matplotlib import pyplot as plt
import matplotlib

H1 = (1/np.sqrt(2)) * np.array([[1,1],[1,-1]])

H2 = np.zeros((4,4))
for i in range(2):
  for j in range(2):
    if i==1 and j==1:
        H2[i*2:i*2+2,j*2:j*2+2] = -H1
    else:
        H2[i*2:i*2+2,j*2:j*2+2] = H1

H2 = (1/np.sqrt(2)) * H2

H3 = np.zeros((8,8))
for i in range(2):
  for j in range(2):
    if i==1 and j==1:
        H3[i*4:i*4+4,j*4:j*4+4] = -H2
    else:
        H3[i*4:i*4+4,j*4:j*4+4] = H2

H3 = (1/np.sqrt(2)) * H3

# order H3
print(H3)

orderlist = []
for i in range(8):
  order = 0
  for j in range(8-1):
    if H3[i,j] * H3[i,j+1] < 0:
      order += 1
  print(order)
  orderlist.append(order)

print("Re-order H3!")
H3_order = np.zeros((8,8))
for i in range(8):
  index = orderlist.index(i)
  print(index)
  H3_order[i] = H3[index]

print("order of ordered H3")
for i in range(8):
  order = 0
  for j in range(8-1):
    if H3_order[i,j] * H3_order[i,j+1] < 0:
      order += 1
  print(order)

# plot H3 basis function



# plot H3 order basis function
linewidth = 1

bf = -0.5 * np.ones((64+ 7*linewidth,64 + 7*linewidth))

for i in range(8):
  for j in range(8):
    bf[i * (8+linewidth): i * (8+linewidth) + 8, j * (8+linewidth): j * (8+linewidth) + 8] = np.outer(H3_order[:,i], H3_order[:,j])

plt.figure()
plt.imshow(bf, cmap='gray')
plt.title('Ordered Hadamard Basis Function')
plt.colorbar()

masked_array = np.ma.masked_where(bf == -0.5, bf)
plt.figure()
cmap = matplotlib.cm.gray  # Can be any colormap that you want after the cm
cmap.set_bad(color='blue')
plt.title('Ordered Hadamard Basis Function for N=8')
plt.imshow(masked_array, cmap=cmap)
plt.axis('off')

# save figure
plt.savefig(work_path+'Plots/Basis_Functions/Ordered.Hadamard.Basis.pdf', format='pdf',
                    bbox_inches='tight', facecolor=(1, 1, 1, 0))

plt.show()
```
